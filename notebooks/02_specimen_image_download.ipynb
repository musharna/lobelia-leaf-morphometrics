```python
import os
import requests

def download_images_from_urls(url_file, output_folder):
    # Create the output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Read URLs from the file
    with open(url_file, 'r') as file:
        urls = file.read().splitlines()

    # Download images
    for i, url in enumerate(urls, start=1):
        try:
            response = requests.get(url, stream=True)
            response.raise_for_status()

            # Extract the file name from the URL
            file_name = os.path.join(output_folder, f'image_{i}.jpg')

            # Save the image
            with open(file_name, 'wb') as image_file:
                for chunk in response.iter_content(chunk_size=8192):
                    image_file.write(chunk)

            print(f"Downloaded image {i}: {url}")
        except requests.exceptions.RequestException as e:
            print(f"Error downloading image {i}: {url}")
            print(f"Error details: {e}")

if __name__ == "__main__":
    # Specify the path to the text file containing URLs
    url_file_path = "C:/Users/a2b32/Downloads/file/media.txt"

    # Specify the output folder for downloaded images
    output_folder_path = "C:/Users/a2b32/Downloads/file/Downloads"

    # Call the function to download images
    download_images_from_urls(url_file_path, output_folder_path)
```

```python
import os
import threading

import pandas as pd
import requests


def download_image(url, filename):
    try:
        response = requests.get(url)
        if response.status_code == 200:
            with open(filename, "wb") as f:
                f.write(response.content)
            print(f"Downloaded: {filename}")
        else:
            print(f"Failed to download: {filename}")
    except Exception as e:
        print(f"An error occurred while downloading {filename}: {e}")


def download_images_multithreaded(dataframe, output_dir, num_threads=5):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    threads = []
    for index, row in dataframe.iterrows():
        url = row["identifier"]
        gbif_id = row["gbifID"]
        filename = os.path.join(output_dir, f"{gbif_id}.jpg")
        thread = threading.Thread(target=download_image, args=(url, filename))
        threads.append(thread)
        if len(threads) >= num_threads:
            for t in threads:
                t.start()
            for t in threads:
                t.join()
            threads = []

    for t in threads:
        t.start()
    for t in threads:
        t.join()


if __name__ == "__main__":
    # Adjust the file path accordingly
    multimedia_file = "C:/Users/<user>/Downloads/0014785-240202131308920/multimedia.txt"

    # Read the Darwin Core file into a DataFrame
    df = pd.read_csv(multimedia_file, sep="\t")

    # Specify the output directory where images will be saved
    output_directory = "downloaded_images"

    # Download images using multi-threading
    download_images_multithreaded(df, output_directory)
```
